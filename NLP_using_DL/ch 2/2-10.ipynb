# 2-10.1 PyKoSpacing

```python
from pykospacing import Spacing
```

PyKoSpacing은 띄어쓰기가 되어있지 않은 문장을 띄어쓰기를 한 문장으로 변환해주는 패키지이다.

```python
sent = '김철수는 극중 두 인격의 사나이 이광수 역을 맡았다. 철수는 한국 유일의 태권도 전승자를 가리는 결전의 날을 앞두고 10년간 함께 훈련한 사형인 유연재(김광수 분)를 찾으러 속세로 내려온 인물이다.'

new_sent = sent.replace(" ", '') # 띄어쓰기가 없는 문장 임의로 만들기
print(new_sent)
```

임의의 문장을 띄어쓰기가 없는 문장으로 만들었다.

```python
spacing = Spacing()
kospacing_sent = spacing(new_sent) 

print(sent)
print(kospacing_sent)
if sent == kospacing_sent:
    print(True)
else:
    print(False)
```

PyKoSpacing 패키지를 사용하여 띄어쓰기를 고친 문장과 원 문장을 비교해보면 결과값이 일치하는 것을 알수있다.

# 2-10.3 SOYNLP 단어 토큰화

soynlp는 품사 태깅, 단어 토큰화를 지원하는 단어 토크나이저이다.  
비지도 학습으로 토큰화를 진행하며, 데이터에 자주 등장하는 단어들을 단어로 분석한다.  
내부적으로 단어 점수 표로 작동하고, 이때 점수는 응집 확률(cohesion probability)와 브랜칭 엔트로피(branching entropy)를 활용한다.

### 1. 신조어 문제

```python
from konlpy.tag import Okt
tokenizer = Okt()
print(tokenizer.morphs('에이비식스 이대휘 1월 최애돌 기부 요정'))
```

기존의 형태소 분석기는 등록되어있지않은 단어들을 구분하지 못한다는 문제점을 가졌다.  
  
위 문장에서 SOYNLP도 제대로 구분되지않은 것처럼 보인다. 하지만 더 많은 텍스트 데이터에서 특정 문자 시퀀스의 빈도가 높고, 앞 뒤로 조사 또는 다른 단어가 등장하는 것도 고려해서 형태소를 판단할 수 있게된다.

### 2. Learning

```python
import urllib.request
from soynlp import DoublespaceLineCorpus
from soynlp.word import WordExtractor
```

```python
urllib.request.urlretrieve("https://raw.githubusercontent.com/lovit/soynlp/master/tutorials/2016-10-20.txt", filename="2016-10-20.txt")
```

```python
# 훈련 데이터를 다수의 문서로 분리
corpus = DoublespaceLineCorpus("2016-10-20.txt")
len(corpus)
```

```python
i = 0
for document in corpus:
  if len(document) > 0:
    print(document)
    i = i+1
  if i == 3:
    break
```

```python
word_extractor = WordExtractor()
word_extractor.train(corpus)
word_score_table = word_extractor.extract() # 단어 점수표 계산
```

SOYNLP는 KoNLPy같은 형태소 분석기와는 달리 학습 과정을 거쳐야한다. 이는 corpus로 응집 확률과 브랜칭 엔트로피 단어 점수표를 만드는 과정이다.
