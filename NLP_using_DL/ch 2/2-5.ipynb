# 2-5.2 정규 표현식 실습

````python
import re
````

search 결과  
- 결과값이 없을 때: 매치되는 문자가 없다.
- 결과값이 존재할 때
   - span=(0, 3): slicing과 같은 의미로 index 0에서 3미만까지를 의미한다.
   - match='abc': 조건에 맞게 매치된 문자의 형태를 알려준다.

### 1) ```.``` 기호

````python
r = re.compile("a.c")
r.search("kkk") # 아무런 결과도 출력되지 않는다.
````

````python
r.search("abc")
````

```.```기호는 임의의 한 문자를 의미하기때문에 a.c는 abc나 axc, a!c같은 모든 형태가 매치된다.

````python
r = re.compile("a..d")
r.search("abcd")
````

```.```기호가 여러개 있어도 개수별 인식이 가능하다.

### 2) ```?```기호

````python
r = re.compile("ab?c")
r.search("abbc") # 아무런 결과도 출력되지 않는다.
````

````python
r.search("abc")
````

```?```앞에 존재하는 단일 문자가 있거나 없어도 된다.

````python
r.search("a1c")
````

문자를 대체하는 것은 불가능하다.

### 3) ```*```기호

````python
r = re.compile("ab*c")
r.search("a") # 아무런 결과도 출력되지 않는다.
````

````python
r.search("ac")
````

````python
r.search("abc") 
````

````python
r.search("abbbbc") 
````

```*```는 앞의 단일 문자가 없거나 무수히많이 존재해도된다.  

````python
r.search("abdc") 
````

````python
r.search("abac") 
````

abdc는 다른 문자가 들어가있어서 매치가 안되지만 abac는 ac가 매치되기때문에 span이 2,4로 매치된다.

### 4) ```+```기호

````python
r = re.compile("ab+c")
r.search("ac") # 아무런 결과도 출력되지 않는다.
````

````python
r.search("abc") 
````

````python
r.search("abbbbc") 
````

```+```는 앞의 단일 문자가 1개 이상이기만하면 매치된다.  
즉, ```*```기호에서 존재하지않는 가정을 없앤 기호이다.

````python
r.search("abac") 
````

```*```기호와는 다르게 abac에서 ac가 매치되지않기때문에 abac도 매치되지않는 문자가 된다.

### 5) ```^```기호

````python
r = re.compile("^ab")

# 아무런 결과도 출력되지 않는다.
r.search("bbc")
r.search("zab")
````

````python
r.search("abz")
````

```^```는 뒤에 따라오는 문자로 시작하는 문자만 매치한다. 그렇기때문에 abz는 매치되지만 zab는 z로 시작하기때문에 매치되지않는다.

### 6) {숫자} 기호

````python
r = re.compile("ab{2}c")

# 아무런 결과도 출력되지 않는다.
r.search("ac")
r.search("abc")
r.search("abbbbbc")
````

````python
r.search("abbc")
````

```{숫자}```는 앞의 단일 문자 하나가 숫자만큼 반복되어있는 문자를 매치한다.

### 7) {숫자1, 숫자2} 기호

````python
r = re.compile("ab{2,8}c")

# 아무런 결과도 출력되지 않는다.
r.search("ac")
r.search("abc")
r.search("abbbbbbbbbc")
````

````python
r.search("abbc")
````

````python
r.search("abbbbbbbbc")
````

```{숫자1, 숫자2}```는 앞의 단일 문자가 숫자1 이상, 숫자2 이하 만큼 반복되어있는 문자를 매치한다.

### 8) {숫자,} 기호

````python
r = re.compile("a{2,}bc")

# 아무런 결과도 출력되지 않는다.
r.search("bc")
r.search("aa")
````

````python
r.search("aabc")
````

````python
r.search("aaaaaaaabc")
````

```{숫자,}```는 앞의 단일 문자가 숫자 이상 만큼 반복되는 문자를 매치한다.  
{0,}은 ```*```과 동일하며, {1,}은 ```+```과 동일한 의미를 가진다.

### 9) [ ] 기호

````python
r = re.compile("[abc]") # [abc]는 [a-c]와 같다.
r.search("zzz") # 아무런 결과도 출력되지 않는다.
````

````python
r.search("a")
````

````python
r.search("aaaaaaa")                                                                                               
````

````python
r.search("baac")      
````

```[ ]``` 안에 문자를 넣으면 그 문자들 중 하나와 매치한다라는 의미이다.

````python
r = re.compile("[a-z]")

# 아무런 결과도 출력되지 않는다.
r.search("AAA")
r.search("111") 
````

````python
r.search("aBC")
````

이런식으로 소문자를 범위 지정하여 매치하는 방법도 존재한다.  
매치 가능한 문자가 여러가지여도 처음에 매치되는 문자만 인식한다.

````python
r = re.compile("[a-z]+")

r.search("abcde")
````

이런 식으로 [조건]뒤에 ```+```를 붙여서 1회 이상을 모두 매치하는 방법을 통해서 더 긴 문자를 매치할 수도 있다.

### 10) [^문자] 기호

````python
r = re.compile("[^abc]")

# 아무런 결과도 출력되지 않는다.
r.search("a")
r.search("ab") 
r.search("b")
````

````python
r.search("d")
````

````python
r.search("1")                                                                                                
````

```[^문자]```는 ```^```기호 뒤에 붙은 문자를 제외한 모든 문자를 매치한다.  
그렇기때문에 숫자를 무시한다면 [^0-9]같은 방식으로 무시하여 매치할 수 있다.

# 2-5.3 정규 표현식 모듈 함수 실습

### 1) re.match()와 re.search()의 차이

````python
r = re.compile("ab.")
r.match("kkkabc") # 아무런 결과도 출력되지 않는다.
````

````python
r.search("kkkabc")  
````

````python
r.match("abckkk")  
````

re.search는 문자의 처음부터 끝까지 정규 표현식과 매치해보는 역할을 한다.  
그에반해 re.match는 문자의 처음부터 매치해보아서 매치되지않으면 그 뒤는 조사하지않는 차이점을 가진다.

### 2) re.split()

````python
# 공백 기준 분리
text = "사과 딸기 수박 메론 바나나"
re.split(" ", text)
````

정규 표현식을 기준으로 문자열을 분리하여 리스트로 만들어준다.  
위와같이 공백을 기준으로 분리를 할수있다.

````python
# 줄바꿈 기준 분리
text = """사과
딸기
수박
메론
바나나"""

re.split("\n", text)
````

````python
# '+'를 기준으로 분리
text = "사과+딸기+수박+메론+바나나"

re.split("\+", text)
````

또한 줄바꿈이나 기호를 기준으로 분리할 수도 있다.

### 3) re.findall()

````python
text = """이름 : 김철수
전화번호 : 010 - 1234 - 1234
나이 : 30
성별 : 남"""

re.findall("\d+", text)
````

````python
re.findall("\d+", "문자열입니다.")
````

정규 표현식과 매치되는 모든 문자들을 리스트로 리턴한다.  
원래는 문자열에서 한 문자씩 분리되지만 ```+```를 추가하면서 연결되어있는 모든 문자열을 읽어온다.  
또한 매치되는 문자가 없으면 빈 리스트를 리턴한다.

### 4) re.sub()

````python
text = "Regular expression : A regular expression, regex or regexp[1] (sometimes called a rational expression)[2][3] is, in theoretical computer science and formal language theory, a sequence of characters that define a search pattern."

preprocessed_text = re.sub('[^a-zA-Z]', ' ', text)
print(preprocessed_text)
````

정규 표현식 패턴과 일치하는 문자를 다른 문자로 대체한다.  
위에서는 특수문자와 숫자를 공백처리하는 용도로 사용한다.

````python
preprocessed_text = re.sub('[^a-z A-Z]', '*', text)
print(preprocessed_text)
````

이와 같이 공백을 추가할수도있고 *로 대체하는 것도 가능하다

# 2-5.4 정규 표현식 텍스트 전처리 실습

````python
text = """100 John    PROF
101 James   STUD
102 Mac   STUD"""

re.split('\s+', text)  #\s+는 공백을 기준으로 모든 문자를 분리한다.
````

````python
re.findall('\d+',text)  #findall()에 \d+를 사용하면 모든 숫자를 찾는다.
````

````python
re.findall('[A-Z]{4}',text)  #대문자가 4번 연속으로 나오는 문자를 찾는다.
````

````python
re.findall('[A-Z][a-z]+',text)  #대분자로 시작하여 소문자가 나오는 문자를 찾는다.
````

# 2-5.5 정규 표현식을 이용한 토큰화

````python
from nltk.tokenize import RegexpTokenizer

text = "Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop"

tokenizer1 = RegexpTokenizer("[\w]+")
tokenizer2 = RegexpTokenizer("\s+", gaps=True)

print(tokenizer1.tokenize(text))
print(tokenizer2.tokenize(text))
````

tokenizer1은 ```\w+```를 사용하였는데 이는 문자 또는 숫자가 1개 이상이라는 의미의 정규 표현식이다.  
tokenizer2는 ```\s+```를 사용하였는데 이는 공백을 찾는 역할이고 gaps를 통해서 이 공백을 토큰으로 삼을지 기준으로 삼을지 결정한다.

````python
tokenizer3 = RegexpTokenizer("\s+", gaps=False)

print(tokenizer3.tokenize(text))
````

예를 들어 gaps=False를 공백을 토큰으로 삼는다는 의미가되기때문에 공백만을 출력하게된다.
