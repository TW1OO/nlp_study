````python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
````

# 2-9.2 x와 y 분리하기

### 1) Split with zip function

````python
X, y = zip(['a', 1], ['b', 2], ['c', 3])
print('X 데이터 :',X)
print('y 데이터 :',y)
````

````python
# 리스트의 리스트 또는 행렬 또는 뒤에서 배울 개념인 2D 텐서.
sequences = [['a', 1], ['b', 2], ['c', 3]]
X, y = zip(*sequences)
print('X 데이터 :',X)
print('y 데이터 :',y)
````

각 배열의 첫번째 원소와 두번째 원소들끼리 묶이는 것을 볼 수 있다.

### 2) Split with DataFrame

````python
values = [['당신에게 드리는 마지막 혜택!', 1],
['내일 뵐 수 있을지 확인 부탁드...', 0],
['도연씨. 잘 지내시죠? 오랜만입...', 0],
['(광고) AI로 주가를 예측할 수 있다!', 1]]
columns = ['메일 본문', '스팸 메일 유무']

df = pd.DataFrame(values, columns=columns)
df
````

````python
X = df['메일 본문']
y = df['스팸 메일 유무']
````

````python
print('X 데이터 :',X.to_list())
print('y 데이터 :',y.to_list())
````

데이터 프레임은 열의 이름으로 접근이 가능하여, 열의 이름으로 x와 y의 데이터를 분리할 수 있다.  

### 3) Split with Numpy

````python
np_array = np.arange(0,16).reshape((4,4))
print('전체 데이터 :')
print(np_array)
````

````python
X = np_array[:, :3]
y = np_array[:,3]

print('X 데이터 :')
print(X)
print('y 데이터 :',y)
````

Numpy에서는 slicing을 사용하여 데이터를 분리할 수 있다.

# 2-9.3 테스트 데이터 분리하기

### 1) Split with sklearn

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state=1234)
```
인자별 의미
- X : 독립 변수 데이터. (배열이나 데이터프레임)
- y : 종속 변수 데이터. 레이블 데이터.
- test_size : 테스트용 데이터 개수를 지정한다. 1보다 작은 실수를 기재할 경우, 비율을 나타낸다.
- train_size : 학습용 데이터의 개수를 지정한다. 1보다 작은 실수를 기재할 경우, 비율을 나타낸다.
- random_state : 난수 시드

````python
# 임의로 X와 y 데이터를 생성
X, y = np.arange(10).reshape((5, 2)), range(5)

print('X 전체 데이터 :')
print(X)
print('y 전체 데이터 :')
print(list(y))
````

````python
# 7:3의 비율로 훈련 데이터와 테스트 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1234)

print('X 훈련 데이터 :')
print(X_train)
print('X 테스트 데이터 :')
print(X_test,'\n')
print('y 훈련 데이터 :')
print(y_train)
print('y 테스트 데이터 :')
print(y_test)
````

train_test_split()을 통해 훈련 데이터와 테스트 데이터를 7:3 비율로 분리하였다.  
결과로 x의 훈련 데이터와 테스트 데이터를 확인해보면 순서가 섞이며 분리된 것을 알 수 있다.

````python
# random_state의 값을 변경
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

print('y 훈련 데이터 :')
print(y_train)
print('y 테스트 데이터 :')
print(y_test)
````

````python
# random_state을 이전의 값이었던 1234로 변경
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1234)

print('y 훈련 데이터 :')
print(y_train)
print('y 테스트 데이터 :')
print(y_test)
````

random_state 값을 변경해보며 결과값이 변화하는 것도 알수있다.  
일반적인 상황에는 random값으로 테스트를 해도되겠지만 동일한 결과를 재현하거나 학습을 할때는 random_state 값을 고정해서 사용할수도 있다.

### 2) 수동으로 분리하기

````python
# 실습을 위해 임의로 X와 y가 이미 분리 된 데이터를 생성
X, y = np.arange(0,24).reshape((12,2)), range(12)

print('X 전체 데이터 :')
print(X)
print('y 전체 데이터 :')
print(list(y))
````

````python
num_of_train = int(len(X) * 0.8) # 데이터의 전체 길이의 80%에 해당하는 길이값을 구한다.
num_of_test = int(len(X) - num_of_train) # 전체 길이에서 80%에 해당하는 길이를 뺀다.
print('훈련 데이터의 크기 :',num_of_train)
print('테스트 데이터의 크기 :',num_of_test)
````

아직 훈련 데이터와 테스트 데이터를 나누지는 않고 각각 몇개의 데이터를 가져갈 것인지 정한 상태이다.  
주의할 점으로 num_of_test의 값을 구할때 ```lne(X) * 0.2```를 사용하면 안된다. 왜냐하면 1234개의 데이터를 가질때 20%의 값은 246.8이며 내림하면 246이고 80%의 값은 987.2로 내림하면 987이다. 두 수를 더하면 1233개로 1개의 데이터가 누락된다.  
그렇기때문에 하나의 값을 먼저 계산하고 그 값을 빼는 방식으로 계산을 해야한다.

````python
X_test = X[num_of_train:] # 전체 데이터 중에서 20%만큼 뒤의 데이터 저장
y_test = y[num_of_train:] # 전체 데이터 중에서 20%만큼 뒤의 데이터 저장
X_train = X[:num_of_train] # 전체 데이터 중에서 80%만큼 앞의 데이터 저장
y_train = y[:num_of_train] # 전체 데이터 중에서 80%만큼 앞의 데이터 저장
````

````python
print('X 테스트 데이터 :')
print(X_test)
print('y 테스트 데이터 :')
print(list(y_test))
````

데이터의 길이는 3으로 잘 나온것을 확인할 수 있지만, 데이터를 섞지않고 어느 지점의 데이터를 일정 길이만큼 분리하였다는 점을 알수있다.  
이 문제를 해결하기위해서는 분리하기 전에 데이터를 수동으로 섞은 후 분리하는 과정이 필요할 수 있다.
