# 2-1.2 토큰화 중 생기는 선택

```python
from nltk.tokenize import word_tokenize
from nltk.tokenize import WordPunctTokenizer
from tensorflow.keras.preprocessing.text import text_to_word_sequence
```

```python
print('단어 토큰화1 :',word_tokenize("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

word_tokenize는 Don't와 Jone's를 **Do/n't**와 **Jone/'s**로 분리하였음을 확인할 수 있다.

```python
print('단어 토큰화2 :',WordPunctTokenizer().tokenize("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

WordPunctTokenizer는 구두점을 별도로 분류하는 특징을 보인다. 즉, Don't와 Jone's를 **Don/'/t**와 **Jone/'/s**로 분류한 것을 확인할 수 있다.

```python
print('단어 토큰화3 :',text_to_word_sequence("Don't be fooled by the dark sounding name, Mr. Jone's Orphanage is as cheery as cheery goes for a pastry shop."))
```

text_to_word_sequence는 케라스의 모델로 모든 알파벳을 lower시킨 후 마침표(.), 느낌표(!) 등의 구두점을 제거한다.  
하지만 apostrophe는 보존하기때문에 don't와 jone's를 그대로 유지하는 모습이다.

# 2-1.3 토큰화에서 고려해야할 사항

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()

text = "Starting a home-based restaurant may be an ideal. it doesn't have a food chain or restaurant of their own."
print('트리뱅크 워드토크나이저 :',tokenizer.tokenize(text))
```

앞에서 말했던 규칙에따라 home-based는 하나의 토큰으로 취급되었으며, doesn't는 does/n't로 분리되었다.

# 2-1.4 Sentence Tokenization

```python
from nltk.tokenize import sent_tokenize

text = "His barber kept his word. But keeping such a huge secret to himself was driving him crazy. Finally, the barber went up a mountain and almost to the edge of a cliff. He dug a hole in the midst of some reeds. He looked about, to make sure no one was near."
print('문장 토큰화1 :',sent_tokenize(text))
```

sent_tokenize는 여러개의 문장에서 문장을 구분하는 코드이다.

```python
text = "I am actively looking for Ph.D. students. and you are a Ph.D student."
print('문장 토큰화2 :',sent_tokenize(text))
```

위의 결과를 보면 단순히 마침표로 구분하여 문장을 구분하지않기때문에 ph.D.를 분리하지않고 단어로 인식하는 것을 볼 수 있다.

```python
import kss

text = '딥 러닝 자연어 처리가 재미있기는 합니다. 그런데 문제는 영어보다 한국어로 할 때 너무 어렵습니다. 이제 해보면 알걸요?'
print('한국어 문장 토큰화 :',kss.split_sentences(text))
```

한국어도 문장 토큰화가 가능한데 이때 KSS를 이용해서 할수있다.

# 2-1.7 Tokenization practice

```python
import nltk
nltk.download('punkt_tab')                      # word_tokenize용
nltk.download('averaged_perceptron_tagger_eng')  # pos_tag용 (버전 이슈로 _eng 접미사 필요)
```

```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

text = "I am actively looking for Ph.D. students. and you are a Ph.D. student."
tokenized_sentence = word_tokenize(text)

print('단어 토큰화 :',tokenized_sentence)
print('품사 태깅 :',pos_tag(tokenized_sentence))
```

태깅 목록
PRP: 인칭 대명사
VBP: 동사
RB: 부사
VBG: 현재부사
IN: 전치사
NNP: 고유 명사
NNS: 복수형 명사
CC: 접속사
DT: 관사

한국어 자연어 처리를 위해서 KoNLPy를 사용하는데 이 안에 형태소 분석기로는 Okt, Mecab, Komoran, Hannanum, Kkma가 존재한다.

```python
from konlpy.tag import Okt
from konlpy.tag import Kkma

okt = Okt()
kkma = Kkma()

print('OKT 형태소 분석 :',okt.morphs("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('OKT 품사 태깅 :',okt.pos("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('OKT 명사 추출 :',okt.nouns("열심히 코딩한 당신, 연휴에는 여행을 가봐요")) 
```

Okt 형태소 분석기 기능
- morphs: 형태소 추출
- pos: 품사 태깅
- nouns: 명사 추출

```python
print('꼬꼬마 형태소 분석 :',kkma.morphs("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('꼬꼬마 품사 태깅 :',kkma.pos("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))
print('꼬꼬마 명사 추출 :',kkma.nouns("열심히 코딩한 당신, 연휴에는 여행을 가봐요"))  
```

Kkma 분석기는 Okt 분석기와 결과가 다르게 나온다.
즉, 결과와 성능이 다르기때문에 분석기의 선택은 필요 용도에 따라 판단한다.
