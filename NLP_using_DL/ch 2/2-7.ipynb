# 2-7.1 Padding with Numpy

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
preprocessed_sentences = [['barber', 'person'], ['barber', 'good', 'person'], ['barber', 'huge', 'person'], ['knew', 'secret'], ['secret', 'kept', 'huge', 'secret'], ['huge', 'secret'], ['barber', 'kept', 'word'], ['barber', 'kept', 'word'], ['barber', 'kept', 'secret'], ['keeping', 'keeping', 'huge', 'secret', 'driving', 'barber', 'crazy'], ['barber', 'went', 'huge', 'mountain']]
```

2-6에서 사용했던 데이터를 그대로 사용한다.

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(preprocessed_sentences)
encoded = tokenizer.texts_to_sequences(preprocessed_sentences)
print(encoded)
```

```python
max_len = max(len(item) for item in encoded)
print('최대 길이 :',max_len)
```

```python
for sentence in encoded:
    while len(sentence) < max_len:
        sentence.append(0)

padded_np = np.array(encoded)
padded_np
```

가장 긴 문장의 길이가 7이었기때문에 모든 문장의 길이를 7로 만들기위해 가상의 단어를 채워서 길이를 맞춘다.  
이때 가상의 단어를 0번으로 정의하는 것을 **zero padding**이라고 한다.

# 2. Padding with Keras

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
encoded = tokenizer.texts_to_sequences(preprocessed_sentences)
print(encoded)
```

```python
padded = pad_sequences(encoded)
padded
```

Numpy로 패딩을 했을때와는 결과가 다른데 그 이유는 Keras의 pad_sequences는 문서의 앞에서부터 0을 채우기 때문이다.

```python
padded = pad_sequences(encoded, padding='post')
padded
```

Numpy와같이 뒤에서 채우고싶으면 pad_sequences를 사용할때 padding 인자에 post를 지정하면된다.

```python
(padded == padded_np).all()
```

Kears와 Numpy 패딩 값이 동일한지 비교한다.

```python
padded = pad_sequences(encoded, padding='post', maxlen=5)
padded
```

실제로는 가장 긴 문서를 기준으로 패딩을 해야만하는 것은 아니다.  
예를 들어 문서의 평균 길이는 20인데 1개의 문서만 길이가 5000이라고 모든 문서의 길이를 5000으로 패딩하는 것은 비효율적이다.  
그렇기때문에 maxlen 인자로 길이 제한을 두고 패딩을 진행할 수도 있다.

```python
padded = pad_sequences(encoded, padding='post', truncating='post', maxlen=5)
padded
```

패딩 길이보다 긴 문서는 데이터 손실이 일어난다.  
이때 truncating인자를 이용해서 데이터 손실이 일어날 위치를 앞이나 뒤로 옮길 수 있다.  
truncating인자를 정의하지않으면 pre로 인식하는데 그러면 앞의 데이터가 손실된다. 반대로 post를 입력하면 뒤의 데이터가 손실된다.

```python
last_value = len(tokenizer.word_index) + 1 # 단어 집합의 크기보다 1 큰 숫자를 사용
print(last_value)
```

```python
padded = pad_sequences(encoded, padding='post', value=last_value)
padded
```

일반적으로 숫자 0으로 패딩하는 것이 관례이긴 하지만, 필수 규칙은 아니다.  
그렇기때문에 위와같이 다른 정수들과 겹치지않도록 집합 크기+1을 숫자로 사용하면 가상의 단어를 14번으로 정의해서 value 인자에 가상의 단어 번호를 입력해서 패딩을 할수있다.
